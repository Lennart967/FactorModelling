**CAPM**

In it's most easy way, a factor model is just a linear combination of different factors to explain one thing. A very basic factor model would be CAPM, which breaks down your portfolio performance into $\beta$, Market Return and $\alpha$

$$
Portfolio Return = \beta * Market Return + \alpha
$$

Beta is the amount of change you can expect in your portfolio when the stock market moves by x. So if your portfolio has a Beta of $\frac{1}{2}$, your portfolio would move by exactly $\frac{1}{2}$. Alpha on the other hand measures your actual skill, if your portfolio has a beta of $\frac{1}{2}$ and your portfolio moved by 0.6 while the market moved by 1, your alpha is 0.1 because in theory, your portfolio should only have moved by 0.5.

$$Example\, of\, CAPM\, on\, AAPL$$

```python
import yfinance as yf
import statsmodels.api as sm

df = yf.download(["AAPL","SPY"], period="max", interval="1d").dropna()
for ticker in ["AAPL", "SPY"]:
    df[("Return", ticker)] = df[("Close", ticker)].pct_change()
    df[("Cumulative Return", ticker)] = (
        (1 + df[("Return", ticker)]).cumprod() - 1
    )
X = df["Close"]["SPY"]
X = sm.add_constant(X)  # -> This will be your Alpha
y = df["Close"]["AAPL"]

model = sm.OLS(y, X).fit()

alpha = model.params["const"]
beta = model.params["SPY"]
print("Linear Regression Results:")
print(f"Alpha: {alpha:.2f}")
print(f"Beta: {beta:.2f}")
print(f"Total AAPL Return: {df['Cumulative Return']['AAPL'].iloc[-1]:.2f}")
print(f"Total SPY Return: {df['Cumulative Return']['SPY'].iloc[-1]:.2f}")
```
